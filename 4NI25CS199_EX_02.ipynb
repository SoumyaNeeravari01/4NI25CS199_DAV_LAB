```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name' : ['Amit','Bebo','Chetan','Amit','Divya','Ravi'],
    'Age' : [25 , np.nan , 30 , 25 , 28 , 22],
    'Marks' : [85 , 90 , np.nan , 85 , 78 , 95],
    'City' : ['Mysuru' , 'Mysuru' , 'Bengaluru' , 'Mysuru', np.nan ,"Delhi"]
}
df=pd.DataFrame(data)
print("Dataset")
df
```

```python
print("\nShape of dataset (rows,columns):", df.shape)
print("\nColumn Name:", df.columns.tolist())
print("\nData type of each columns:")
print(df.dtypes)
print("\nData info:")
df.info()
```

```python
print("\n descriptive stastitics(numerical columns):")
print(df.describe())
print("\nMean:\n",df.mean(numeric_only=True))
print("\nMedian:\n",df.median(numeric_only=True))
print("\nStandard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\nMissing values in each columns:")
print(df.isnull().sum())
print("\nTotal missing values in dataset:", df.isnull().sum().sum())
```

```python
print("\n Duplicate rows in dataset:")
print(df[df.duplicated()])
print("\nNumber of duplicate rows:",df.duplicated().sum())
print("\nEDA process completed successfully!")
```
