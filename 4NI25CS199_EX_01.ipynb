```python
data = {
    "Name": ["Ahsa","Bahart","Chetan","Divya","Esha"],
    "Branch": ["CSE","ISE","CSE","AIML","ISE"],
    "Marks":[72,91,84,95,68],
    "Attendance":[82,94,88,96,75]
}
students = pd.DataFrame(data)
students
```

```python
print(students["Name"])
print(students["Marks"])
```

```python
print("student scoring more than 80:")
display(students[students["Marks"]>80])
```

```python
print("Student attendance more than 85 percentage:")
display(students[students["Attendance"]>85])
```

```python
print("First three Students :\n",students[0:3])
```

```python
print("Average of marks:",students["Marks"].mean())
```

```python
print("Maximum Marks:",students["Marks"].max())
```

```python
print("Students from only CSE Branch:",students[students["Branch"]=="CSE"])
```

```python
print("Student scored atleast 80 and having tne attendance 85 percentage:")
display(students[(students["Marks"]>=80) & (students["Attendance"]>85)])
```
